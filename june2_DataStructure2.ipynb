```python
l = [34, 56, 12, 48, 19, 34]
```

```python
i = 0 
for n in l :
    print(i , "=>",n)
    i+=1                      
```

```python
for i in l:
    print(f"{l.index(i)} => {i}")  # give same index to duplicate value
```

```python
list(enumerate(l))
```

```python
for ind , val in enumerate(l):
    print(f"{ind} => {val}")
```

```python
for i in range(len(l)):
    print(f"{i} => {l[i]}")   # using length function
```

```python
list(enumerate(l, start= 5)) # we can give the starting index
```
