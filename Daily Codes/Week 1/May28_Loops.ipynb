```python
no = 5
fact = 1
while no != 0:
    fact = fact * no 
    no -= 1
print("factorial :" , fact)
```

```python
no = int (input("enter the number:"))

fact = 1

while no != 0:
    fact = fact * no 
    no-=1
print("factorial:", fact)
```

```python
num = int(input("enter the number "))

digit_sum = 0

while num > 0:
    digit = num % 10
    digit_sum = digit_sum  + digit 
    num = num // 10

print("addition of digit:", digit_sum)
```

```python
# break and continue
```

```python
for i in range (1 , 11):
    if i==5:
        break
    print(i)
    print('ok')
```

```python
for i in range (1, 11):
    print(i)
    print('ok')
    if i == 5:
        break
```

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```
