## inheritance 

```python
class Person:
    insti = "MITU"

    def __init__(self , age , ht):
        self.age = age 
        self.height = ht

    def show(self):
        print(f" Age =>{self.age} , Height =>{self.height} , Institude =>{self.insti}")
        
```

```python
class Ravi(Person):

    def __init__(self , age , ht , sal) :
        super().__init__(age , ht)
        self.sal = sal

    def show_sal(self):
        print(f"Salary =>{self.sal}")
```

```python
a = Ravi(25 , 6.1 , 60000)
```

```python
a.show_sal()
```

```python
a.show()
```

```python
class Menuapp:

    def __init__(self) :
        self.itemlist = [] 

    def show_menu(self):
        print("Menu \n 1.add item\2. view items \3.exit")

    def add_items(self):
        item = input("Enter the item ")
        self.itemlist.append(item)

    def view_item(self):
        print(self.itemlist)

    def run(self):
        
        while True :
            self.show_menu()
            ch = int(input("Enter the choice"))

            if ch == 1 :
                self.add_items()
            elif ch == 2 :
                self.view_item()
            elif ch == 3:
                break
            else:
                return "invalid choice"
                
```

```python
m = Menuapp()
```

```python
m.run()
```

```python
class Person :

    def __init__(self , age , name):
        self.__age = age
        self.__name = name

    #geter 
    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

    #setter
    def set_age(self , age):
        self.__age = age

    def __str__(self):
        return f"age=> {self.__age} , name => {self.__name}"
```

```python
class Bankapp:

    def __init__(self , name , id , Balance = 0 ) :

    def show_menu(self):
        print("Menu \n1.check balance \n2.deposit money \n3.withdraw money \n4.exit")

    def check_balance(self):
        print("Current balance :" ,self.Balance)

    def deposit_money(self):
        amount = float(input("enter the amount:"))
        self.Balance += amount
            
    def withdraw_money(self):
        amount = int(input("enter the amount:"))
        if amount <= self.Balance:
            self.Balance -= amount
        else:
            print("uniff")

    def run(self):
        
        while True :
            self.show_menu()
            ch = int(input("Enter the choice"))

            if ch == 1 :
                self.check_balance()
            elif ch == 2 :
                self.deposit_money()
            elif ch == 3:
                self.withdraw_money()
            elif ch == 4:
                break
            else:
                return "invalid choice"
```
