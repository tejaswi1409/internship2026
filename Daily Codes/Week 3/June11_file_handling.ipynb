```python
# write()
```

```python
sent = "frod took the ring .he started his journey to mount doom. "
```

```python
f = open("myfile.txt" , "w")
f.write(sent)
f.close()
```

```python
sent = "frod took the ring. \n he started his journey to mount doom. "
```

```python
with open ("myfile.txt" , "w") as f:
    f.write(sent)
```

```python
# simple file copy operation
```

```python
s = open('sample.txt' , 'r')
f = open('myfile.txt' , 'w')
```

```python
contents = s.read()
f.write(contents)
f.close()
```
