```python
class Person:

    def __init__(self , name ,age, hg):
        self.name = name
        self.age = age  
        self.height = hg

    def  show_info(self):
        return f"name => {self.name} age => {self.age} height => {self.height}"
        
```

```python
p1 = Person("sai" , 21 , 6.1)
p2 = Person("raj" , 22 , 5.11)
```

```python
p1.show_info()
```

```python
class Person:

    def __init__(self , name ,age, hg):
        self.name = name
        self.__age = age   # private variable (__)
        self.height = hg

    def  show_info(self):
        return f"name => {self.name} age => {self.__age} height => {self.height}"
```

```python
p1 = Person("sai" , 21 , 6.1)
p2 = Person("raj" , 22 , 5.11)
```

```python
p1.show_info()
```

```python
class Employee:

    def __init__(self , company , name , sal , desi):
        self.company = company
        self.name = name
        self.__salary = sal
        self.designation = desi

    def calculate(self):
        return f"Annual salary of {self.name} :{self.__salary * 12}"

    def show_info(self):
        return f"Company name :{self.company} , Employee name :{self.name} ,  Salary :{self.__salary} , Designation :{self.designation}"
        
```

```python
e1 = Employee("Tata" , "Mohit" , 85000 , "manager")
e2 = Employee("Tata" , "Rohan" , 79000 , "Software engineer")
e3 = Employee("Tata" , "Aditya" , 90000 , "sr.manager")
```

```python
e1.show_info()
```

```python
e1.calculate()
```
