```python
try :
    f = open('a1.txt')
    print(var1)
except FileNotFoundError:
    print("file does not exist")
```

```python
try :
    f = open('a1.txt')
    print(var1)
except FileNotFoundError:
    print("file does not exist")
except Exception:
    print("something went wrong")
```

```python
try :
    f = open('a1.txt')
    print(var1)
except FileNotFoundError:
    print("file does not exist")
except Exception as e:
    print(e)
```

```python
# else -> if try block doesnt raise any exception this eill run 
```

```python
try:
    f = open("a1.txt")
except FileNotFoundError:
    print("file does not exist")
except Exception as e:
    print(e) 
else:
    print(f.read())
    f.close
```

```python
try:
    f = open("a1.txt")
    if f.name == 'a1.txt':
        raise Exception
except FileNotFoundError:
    print("file does not exist")
except Exception as e:
    print(e) 
else:
    print(f.read())
    f.close
finally:
    print("run complete")
```

```python
try:
    a = int(input("enter the number 1:"))
    b = int(input("enter the number 2:"))
    div = a / b
except ZeroDivisionError:
    print("number cannot divide by zero ")
except ValueError:
    print("the value is wrong")
else:
    print("Division:" , div)
finally:
    print("run complete")
```
