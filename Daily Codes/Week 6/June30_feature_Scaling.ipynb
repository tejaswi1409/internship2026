```python
df = pd.read_csv("Social_Network_Ads.csv")
```

```python
plt.figure(figsize=(16,9))
plt.scatter(df["Age"],df["EstimatedSalary"],c=df["Purchased"]);
plt.xlabel("Age")
plt.ylabel("Estimated Salary")
plt.title("Data Distribution")
```
