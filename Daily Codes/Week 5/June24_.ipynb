```python
def Remove_outlier(df,Age):
    Q1 = df[Age].quantile(0.25)
    Q3 = df[Age].quantile(0.75)
    IQR = Q3 - Q1 

    UB = Q3 + 1.5 * IQR
    LB = Q1 - 1.5 * IQR

    df = df[(df[Age] >= LB)  & (df[Age] <= UB)]
    return df

```

```python
df = Remove_outlier(df,'Age')
```
