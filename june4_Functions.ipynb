```python
def large(x,y):
    
    if x > y :
        print("x is larger than y")
    elif y >x :
        print("y is larger than x")
    else:
        print("both are equal")
```

```python
large(10,32)
```

```python
large(10,10)
```

```python
def display(name):
    i = 1
    while i <= 4 :
        print(name)
        i+=1
```

```python
def dis(name,n):
    for i in range(n):
        print(name)
```

```python
# function to find factorial of number
```

```python
def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact = fact*i
    print(fact)
```

```python
factorial(5)
```

```python
list(n2)
```
