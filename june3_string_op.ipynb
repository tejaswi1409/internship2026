```python
sent = input("Enter the string :")
ch1 = input("Enter the character 1:")
ch2 = input("Enter the character 2:")

for i in [ch1 , ch2]:
    c = False
    for j in sent:
        if i==j:
            c = True
    if c :
        print(i,"is present")
    else:
        print(i,"is not present")
```

```python
sent = input("Enter the string :")
d = {'a':0 , 'e':0,'i':0, 'o':0, 'u':0}

count = 0

for i in  sent.lower():
    if i in d :
        d[i] = d[i] + 1
        count = count + 1

print("total vowels :", count)
print("each vowel count: ", d)
   
```
