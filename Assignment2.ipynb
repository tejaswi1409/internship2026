```python
def analyze_sentence(sentence) :
    words = sentence.lower().split()
    
    result ={
        "word_count": len(words),
        "char_count": len(sentence),
        "unique_words": len(set(words)),
        "most_frequent_word": max(set(words), key=words.count)
    }

    print(result)
```

```python
analyze_sentence("The quick brown fox jumps over the lazy dog the fox")
```

```python
# a) Reverse the string without using slicing

text = "A man a plan a canal Panama"

rev =""
for i in text:
    rev = i + rev 
print("Reverse :",rev)
```

```python
# b) Check if it is a palindrome, ignoring case and spaces.

text = "A man a plan a canal Panama"

clear = text.replace(" ","").lower()

rev =""
for i in clear:
    rev = i + rev

if clear == rev:
    print("The string is palindrome ")
else:
    print("The string is not palindrome")
```

```python
# c) Replace every vowel with *.

text = "A man a plan a canal Panama"

new_str = ""

vowels = {"a","e" ,"i","o" ,"u"}

for i in text:
    if i.lower() in vowels:
        new_str +="*"
    else:
        new_str += i

print("Replace :", new_str)
```

```python
# d) Count how many times each vowel appears.

text = "A man a plan a canal Panama"

vowels = {
    "a" : 0 ,
    "e" : 0 ,
    "i" : 0 ,
    "o" : 0 ,
    "u" : 0 , 
}

for i in text.lower():
    if i in vowels :
        vowels[i] +=1

print("vowel count :" , vowels)
```

```python
def inventory_manager(records):
    result ={}
    for product , category , quantity , price in records:
        if category not in result :
            result[category] ={
                "total_items" : 0 , 
                
                "total_value" : 0 ,
                    
                "product" : []
            }
        result[category]["total_items"] +=quantity
        
        result[category]["total_value"] +=quantity * price
        
        result[category]["product"].append(product)

    return result
```

```python
inventory_manager([ 
("Laptop", "Electronics", 10, 55000),
("Mouse", "Electronics", 50, 499), 
("Desk", "Furniture", 8, 12000), 
("Chair", "Furniture", 15, 4500),
("Pen", "Stationery", 200, 20), ])
```

```python
# a) A function power_all(*args, exp=2) that returns a list of each number raised to exp.

def power_all(*args, exp=2):
    result =[]
    for i in args:
        result.append(i ** exp)
    return result 
```

```python
# a) Write a closure make_multiplier(n) that returns a function multiplying its input by n. Create triple and quintuple using it.

def make_multiplier(n):
    def multiply(x):
        return x * n
    return multiply


triple = make_multiplier(3)
quintuple = make_multiplier(5)

print(triple(10))     
print(quintuple(10))  
```

```python
# b) Write a decorator @timer that prints the execution time of any function it wraps.
# Apply it to slow_sum(n) which computes the sum of squares from 1 to n

import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print("Execution time:", end - start, "seconds")
        return result
    return wrapper
```

```python
@timer
def slow_sum(n):
    return sum(i * i for i in range(1, n + 1))


print(slow_sum(100000))
```

```python
# c) Write a decorator @validate_positive that raises a ValueError if any argument is zero or negative. Apply it to safe_sqrt(x)

def validate_positive(func):
    def wrapper(x):
        if x <= 0:
            raise ValueError("Input must be positive")
        return func(x)
    return wrapper
```

```python
import math

@validate_positive
def safe_sqrt(x):
    return math.sqrt(x)


print(safe_sqrt(25))  
```
