```python
nums = int(input("Enter the numbers until enter 0 "))

count = 0
sum = 0

while nums != 0 :
    sum = sum + nums
    count = count +1
    nums = int(input("Enter the numbers until enter 0 : "))
    
print("sum of numbers :" , sum)
print("count of numbers : ", count)

if count > 0 :
    print("Average :" , sum / count)
```
