Create two BankAccount objects and perform a few deposits and withdrawals on each

```python
class BankAccount:
    def __init__(self ,account_holder , account_number , balance ):
        self.account_holder = account_holder 
        self.account_number = account_number
        self.balance = balance

    def deposit(self , amount):
        self.balance += amount
        print("money succesfully deposited ")

    def withdraw(self , amount):
        if self.balance < amount:
            print("Inefficient ammount")
        else :
            self.balance -= amount
            print("money withdraw from account successfully")

    def __str__(self):
        return f"Account Holder => {self.account_holder} , Account Number => {self.account_number} , Balance => {self.balance}  "

    
```

2. Create a class Student with attributes name, roll_no, and marks (a list of 5 subject marks). Add methods: a) calculate_percentage() — returns the average of all marks. b) get_grade() — returns 'A' for >= 80, 'B' for >= 60, 'C' for >= 40, and 'Fail' otherwise. c) display() — prints all details including percentage and grade. Create a list of 3 Student objects and call display() on each

```python
class Student:
    def __init__(self , name , roll_no , marks):
        self.name = name 
        self.roll_no = roll_no
        self.marks = marks

    def calculate_percentage(self):
        return sum(self.marks) / len(self.marks)

    def get_grade(self):

        per = self.calculate_percentage()

        if per >=80:
            return " A "
        elif per >=60:
            return " B "
        elif per>=40:
            return " C "
        else:
            return "Fail"

    def display(self):
        print("Name =>" ,self.name)
        print("Roll no => " ,self.roll_no)
        print("Marks =>" ,self.marks)
        print("Percentage => ",self.calculate_percentage())
        print("Grade => ", self.get_grade())
```

5. Create a base class Shape with a method area() that returns 0 and a method describe() that prints 'I am a shape'. Create three subclasses: a) Circle(radius) — overrides area() using pi * r^2. b) Rectangle(length, width) — overrides area(). c) Triangle(base, height) — overrides area(). Each subclass should also override describe() to print something specific about itself. Create one object of each and call both methods

```python
class Shape:
    def area():
        return 0

    def describe():
        print("i am a shape")
```

```python
import math

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return f"Area of circle : {math.pi * self.radius ** 2}"

    def describe(self):
        print("i am a circle ")

class Rectangle(Shape):
    def __init__(self , lenght , width):
        self.lenght = lenght 
        self.width = width 

    def area(self) :
        return f"Area of Rectangle :{self.lenght * self.width}"

    def describe(self):
        print("i am a rectangle ")

class Triangle(Shape):
    def __init__(self , base , height):
        self.base = base 
        self.height = height

    def area(self):
        return f"Area of Triangle : {0.5 * self.base  * self.height}"

    def describe(self):
        print("i am a triangle")
        
```

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def start_engine(self):
        print("Engine started")
```

```python
class Car(Vehicle):
    def __init__(self, make, model, year, num_doors):
        super().__init__(make, model, year)
        self.num_doors = num_doors

    def start_engine(self):
        print(f"Engine started for {self.make} ")

class ElectricCar(Vehicle):
    def __init__(self, make, model, year, battery_capacity):
        super().__init__(make, model, year)
        self.battery_capacity = battery_capacity

    def start_engine(self):
        print(f"Silent start — battery at {self.battery_capacity} kWh")
```

```python
# b) Read the file back and print each name with its line number.

print(" Students with line numbers:")

with open ("students.txt", "r") as f:
    
    for line_no, name in enumerate(f, start=1):
        print(f"{line_no}. {name}")
```

10. Build a menu-driven Student Record Manager using OOP and file handling. The app should have a class StudentManager with the following menu options: 1 . Add Student — takes name, roll number, and marks for 3 subjects. 2 . View All — displays all stored students with their average marks. 3 . Search Student — finds a student by roll number. 4 . Delete Student — removes a student by roll number. 5 . Save to File — writes all records to students_data.txt. 6 . Load from File — reads records back from students_data.txt on startup. 7 . Exit

```python
class StudentManager:
    def __init__(self):
        self.students = []
        

    def add_student(self):
        roll = input("Enter Roll Number: ")
        name = input("Enter Name: ")

        marks1 = float(input("Enter Marks for Subject 1: "))
        marks2 = float(input("Enter Marks for Subject 2: "))
        marks3 = float(input("Enter Marks for Subject 3: "))

        s = {
            "roll": roll,
            "name": name,
            "marks": [marks1, marks2, marks3]
        }

        self.students.append(s)
        print("Student added successfully!")

    def view_all(self):

        print("Student Details :")

        for s in self.students:
            avg = sum(s["marks"]) / 3

            print(f"Roll No : {s['roll']}")
            print(f"Name    : {s['name']}")
            print(f"Marks   : {s['marks']}")
            print(f"Average : {avg} \n " )

    def search_student(self):
        roll = input("Enter Roll Number to Search: ")

        for s in self.students:
            if s["roll"] == roll:
                
                avg = sum(s["marks"]) / 3
        
                print("Student Found :")
                 
                print(f"Roll No : {s['roll']}")
                print(f"Name    : {s['name']}")
                print(f"Marks   : {s['marks']}")
                print(f"Average : {avg} \n ")
                return
                

        print("Student not found.")

    def delete_student(self):
        roll = input("Enter Roll Number to Delete: ")

        for s in self.students:
            if s["roll"] == roll:
                self.students.remove(s)
                print("Student deleted successfully!")
                

        print("Student not found.")

    def save_to_file(self):
        f = open("students_data.txt", "w")
        for s in self.students:
            line = (
                f"{s['roll']},"
                f"{s['name']},"
                f"{s['marks'][0]},"
                f"{s['marks'][1]},"
                f"{s['marks'][2]}"
                )
            f.write(line)


    def load_from_file(self):
        f = open ("students_data.txt", "r")
        for line in f:
            print(line.strip())
            
    
    def run(self):
        while True:
            
            print(" Menu :")
            print("1. Add Student")
            print("2. View All")
            print("3. Search Student")
            print("4. Delete Student")
            print("5. Save to File")
            print("6 . Load from file")
            print("6. Exit \n ")

            ch = input("Enter the choice: ")

            if ch == "1":
                self.add_student()

            elif ch == "2":
                self.view_all()

            elif ch== "3":
                self.search_student()

            elif ch == "4":
                self.delete_student()

            elif ch == "5":
                self.save_to_file()

            elif ch == "6":
                self.load_from_file()
                
            elif ch == "7":
                print("Exit")
                break
              

            else:
                print("Invalid choice")
```
